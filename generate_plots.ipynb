```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("pickxtimes_bc_training.csv")
df = df.sort_values("Step")

df = df[df["Step"] >= 1]

x = df["Step"]

plt.figure(figsize=(10, 6))

# -------------------------
# Train (blue family)
# -------------------------
plt.plot(x, df["train.loss"],        color="#1f77b4", label="train.loss", linewidth=2)
plt.plot(x, df["train.ptp_loss"],     color="#4fa3dd", label="train.ptp_loss", linestyle="--")
plt.plot(x, df["train.action_loss"],  color="#0b3c6f", label="train.action_loss", linestyle=":")

# -------------------------
# Val (orange family)
# -------------------------
plt.plot(x, df["val.loss"],        color="#ff7f0e", label="val.loss", linewidth=2)
plt.plot(x, df["val.ptp_loss"],    color="#ffb570", label="val.ptp_loss", linestyle="--")
plt.plot(x, df["val.action_loss"], color="#a65300", label="val.action_loss", linestyle=":")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss Curves")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
from PIL import Image
import math

# ==== SETTINGS ====
POSTER_WIDTH_INCH  = 24
POSTER_HEIGHT_INCH = 36
DPI = 300

A4_WIDTH_INCH  = 8.5
A4_HEIGHT_INCH = 11

# Convert dimensions to pixels
POSTER_W = round(POSTER_WIDTH_INCH * DPI)
POSTER_H = round(POSTER_HEIGHT_INCH * DPI)

A4_W = round(A4_WIDTH_INCH * DPI)
A4_H = round(A4_HEIGHT_INCH * DPI)

# ==== LOAD IMAGE ====
Image.MAX_IMAGE_PIXELS = 1_000_000_000
img = Image.open("CS224R_Poster.jpg")

# ==== RESIZE TO POSTER SIZE ====
img = img.resize((POSTER_W, POSTER_H), Image.LANCZOS)

# ==== CALCULATE GRID ====
cols = math.ceil(POSTER_W / A4_W)
rows = math.ceil(POSTER_H / A4_H)

print(f"Poster size: {POSTER_WIDTH_INCH}\" × {POSTER_HEIGHT_INCH}\"")
print(f"Resolution: {POSTER_W} × {POSTER_H} pixels")
print(f"Slicing into {cols} × {rows} A4 pages ({cols * rows} sheets)")

# ==== SLICE INTO A4 TILES ====
index = 1
for r in range(rows):
    for c in range(cols):
        left = c * A4_W
        top = r * A4_H
        right = min(left + A4_W, POSTER_W)
        bottom = min(top + A4_H, POSTER_H)

        tile = img.crop((left, top, right, bottom))
        tile.save(
            f"poster_A4_{index:02d}.jpg",
            dpi=(DPI, DPI),
            quality=95
        )
        print(f"Saved poster_A4_{index:02d}.jpg")

        index += 1

print("\nDone! Print the A4 pages at 100% scale and assemble them.")
```

```python
from huggingface_hub import hf_hub_download
import zipfile
import io
import torch

repo_id = "Yinpei/mme_vla_suite"
model_name = "perceptual-tokendrop-modul"

# --------------------------------------------------
# 1. Download ZIP (HF handles caching automatically)
# --------------------------------------------------
zip_path = hf_hub_download(
    repo_id=repo_id,
    filename=f"{model_name}/79999.zip"
)

print("[OK] ZIP cached at:", zip_path)

# --------------------------------------------------
# 2. Inspect ZIP WITHOUT extracting
# --------------------------------------------------
with zipfile.ZipFile(zip_path, "r") as z:
    files = z.namelist()
    print("\n[ZIP sample files]")
    print(files[:20])

    ckpt_files = [f for f in files if f.endswith((".pt", ".pth", ".ckpt", ".bin"))]
    print("\n[checkpoint candidates]")
    print(ckpt_files)

# --------------------------------------------------
# 3. Load checkpoint IN MEMORY (no disk usage)
# --------------------------------------------------
if len(ckpt_files) > 0:
    ckpt_file = ckpt_files[0]

    with zipfile.ZipFile(zip_path, "r") as z:
        with z.open(ckpt_file) as f:
            buffer = io.BytesIO(f.read())
            ckpt = torch.load(buffer, map_location="cpu")

    print("\n[OK] checkpoint loaded in memory")
    print("type:", type(ckpt))

    if isinstance(ckpt, dict):
        print("keys:", list(ckpt.keys())[:20])
```
